```python
import random

jobs = [2, 5, 3, 7, 4, 6, 8, 1]
machines = 3
population_size = 10
generations = 30
mutation_rate = 0.1

def fitness(solution):
    load = [0] * machines
    for i, job in enumerate(jobs):
        load[solution[i]] += job
    return max(load)

def create_population():
    return [
        [random.randint(0, machines - 1) for _ in jobs]
        for _ in range(population_size)
    ]

def selection(population):
    selected = []
    for _ in range(population_size):
        a, b = random.sample(population, 2)
        selected.append(a.copy() if fitness(a) < fitness(b) else b.copy())
    return selected

def crossover(p1, p2):
    point = random.randint(1, len(jobs) - 1)
    return p1[:point] + p2[point:], p2[:point] + p1[point:]

def mutation(solution):
    for i in range(len(solution)):
        if random.random() < mutation_rate:
            solution[i] = random.randint(0, machines - 1)
    return solution

population = create_population()
best = None

for generation in range(generations):
    selected = selection(population)
    new_population = []

    for i in range(0, population_size, 2):
        p1 = selected[i]
        p2 = selected[(i + 1) % population_size]

        c1, c2 = crossover(p1, p2)

        new_population.append(mutation(c1))
        new_population.append(mutation(c2))

    population = new_population

    current_best = min(population, key=fitness)

    if best is None or fitness(current_best) < fitness(best):
        best = current_best.copy()

print("Best Job Assignment:")

for machine in range(machines):
    assigned = [jobs[i] for i in range(len(jobs)) if best[i] == machine]
    print("Machine", machine + 1, ":", assigned)

print("Minimum Makespan:", fitness(best))
```
