```python
HEALTHY = 0
BURNING = 1
BURNED = 2

size = 5
forest = [[HEALTHY for _ in range(size)] for _ in range(size)]

forest[2][2] = BURNING

def display_forest(forest):
    symbols = {HEALTHY: "T", BURNING: "F", BURNED: "X"}
    for row in forest:
        print(" ".join(symbols[cell] for cell in row))
    print()

def update_forest(forest):
    new_forest = [row[:] for row in forest]

    for i in range(size):
        for j in range(size):
            if forest[i][j] == BURNING:
                new_forest[i][j] = BURNED

                for di in range(-1, 2):
                    for dj in range(-1, 2):
                        ni = i + di
                        nj = j + dj

                        if 0 <= ni < size and 0 <= nj < size:
                            if forest[ni][nj] == HEALTHY:
                                new_forest[ni][nj] = BURNING

    return new_forest

iteration = 1

while any(BURNING in row for row in forest):
    print("Iteration:", iteration)
    display_forest(forest)

    forest = update_forest(forest)
    iteration += 1

print("Final Forest:")
display_forest(forest)
```
